```python
import pandas as pd
customers = pd.read_csv("03_Library SystemCustomers.csv")#
books = pd.read_csv("03_Library Systembook.csv")

customers.head()
```

```python
books.head()
```

```python
customers = customers.dropna()
```

```python
books.dropna()
```

```python
books = books[books["Customer ID"].isin(customers["Customer ID"])]
```

```python
display(customers)
display(books)
```

```python
customers.to_csv("clean_customers.csv", index = False)
books.to_csv("clean_books.csv", index = False)
```

```python
books["Book checkout"] = books["Book checkout"].str.replace('"','')
books["Book Returned"] = books["Book Returned"].str.replace('"','')
```

```python
display(customers)
display(books)
```

```python
books["Book checkout"] = pd.to_datetime(books["Book checkout"], dayfirst=True, errors="coerce")
books["Book Returned"] = pd.to_datetime(books["Book Returned"], dayfirst=True, errors="coerce")
```

```python
books = books.dropna()
```

```python
books["days Borrowed"] = (books["Book Returned"] - books["Book checkout"]).dt.days
```
